```python
import pandas as pd
data_name='planning_datasets/merged_train.jsonl'
data = pd.read_json(data_name, lines=True)
```

```python
data['clean_question'] = None
for idx, row in data.iterrows():
    if 'meeting' in row['index']:
        # print(row['prompt_0shot'])
        data.loc[idx, 'clean_question'] = row['prompt_0shot'].replace("Your response should start with 'SOLUTION:'.", "Generate a plan for meeting as much friends as possible within the constraints. Plan step by step, each step should be a part of the meeting plan.")
    elif 'trip' in row['index']:
        data.loc[idx, 'clean_question'] = row['prompt_0shot']+ "\nGenerate a plan for the trip that adheres to the constraints. Plan step by step, each step should be a part of the trip plan."
    elif 'calendar' in row['index']:
        data.loc[idx, 'clean_question'] = row['prompt_0shot'] + "Reason step by step to arrive at the final proposed time."
    else:
        raise ValueError(f"Unknown dataset: {row['index']}")
```
