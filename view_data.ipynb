```python
import pandas as pd
data_name='planning_datasets/merged_train.jsonl'
data = pd.read_json(data_name, lines=True)
```

```python
data['clean_question'] = None
for idx, row in data.iterrows():
    if 'meeting' in row['index']:
        # print(row['prompt_0shot'])
        data.loc[idx, 'clean_question'] = row['prompt_0shot'].replace("Your response should start with 'SOLUTION:'.", "Generate a plan for meeting as much friends as possible within the constraints. Plan step by step, each step should be a part of the meeting plan.")
    elif 'trip' in row['index']:
        data.loc[idx, 'clean_question'] = row['prompt_0shot']+ "\nGenerate a plan for the trip that adheres to the constraints. Plan step by step, each step should be a part of the trip plan."
    elif 'calendar' in row['index']:
        data.loc[idx, 'clean_question'] = row['prompt_0shot'] + "Reason step by step to arrive at the final proposed time."
    else:
        raise ValueError(f"Unknown dataset: {row['index']}")
```

```python
# Find common columns between data and full_data
common_cols = list(set(data.columns) & set(full_data.columns))
print("Common columns:", common_cols)

# Create a function to generate a hash of row values for matching
def get_row_hash(row, cols):
    return hash(tuple(str(row[col]) for col in cols))

# Add hash values to both dataframes
data['row_hash'] = data.apply(lambda row: get_row_hash(row, common_cols), axis=1)
full_data['row_hash'] = full_data.apply(lambda row: get_row_hash(row, common_cols), axis=1)

# Find matching indices
matches = []
for idx, row in data.iterrows():
    matching_indices = full_data[full_data['row_hash'] == row['row_hash']].index.tolist()
    if matching_indices:
        matches.append((idx, matching_indices[0]))

print("\nFound", len(matches), "matching rows")
print("\nFirst 10 matches (data_index, full_data_index):")
for match in matches[:10]:
    print(match)

# Clean up temporary columns
data.drop('row_hash', axis=1, inplace=True)
full_data.drop('row_hash', axis=1, inplace=True)
```

```python
import pandas as pd
from datasets import Dataset, concatenate_datasets, load_from_disk

# Load both datasets
data1 = load_from_disk('train_naturalplan_Qwen3-8B')
data2 = load_from_disk('train_naturalplan_Qwen3-8B_part2')

# Convert pandas DataFrames to datasets
if isinstance(data1, pd.DataFrame):
    data1 = Dataset.from_pandas(data1)
if isinstance(data2, pd.DataFrame):
    data2 = Dataset.from_pandas(data2)

# Concatenate the datasets
combined_data = concatenate_datasets([data1, data2])

# Save the combined dataset
combined_data.save_to_disk('train_naturalplan_Qwen3-8B_combined')

print(f"Combined dataset saved with {len(combined_data)} examples")
```
